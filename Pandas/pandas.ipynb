## Resources

**Pandas** is a fast, powerful, flexible, and easy-to-use open-source data analysis and manipulation tool, built on top of the Python programming language. It is particularly well-suited for working with tabular data (like spreadsheets or SQL tables) and time series data. Its two primary data structures are `DataFrame` and `Series`.

```python
info = {
    "Name" : ["Rahule","Harshita" , "Shradha"],
    "CGPA" : [ 9.5,7.5,8.5]
    }

df = pd.DataFrame(info)
print(df)
```
